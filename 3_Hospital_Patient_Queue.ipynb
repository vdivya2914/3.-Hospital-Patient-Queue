```python
import heapq

patients = []

while True:
    print("\n1. Add Patient")
    print("2. Treat Patient")
    print("3. Show Patients")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Patient Name: ")
        priority = int(input("Priority (1=Emergency, 2=Normal, 3=Low): "))

        heapq.heappush(patients, (priority, name))
        print("Patient added!")

    elif choice == "2":
        if patients:
            priority, name = heapq.heappop(patients)
            print("Treating:", name)
        else:
            print("No patients.")

    elif choice == "3":
        print("Waiting Patients:")

        for patient in sorted(patients):
            print(patient[1], "- Priority", patient[0])

    elif choice == "4":
        break

    else:
        print("Invalid choice")
```
